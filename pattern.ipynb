```python
n = 5

# Upper part including middle
for i in range(n):
    print("" * i, end="")

    start = 2 * n - 2 - i

    for j in range(2 * n - 1 - 2 * i):
        print(chr(65 + start - j), end=" ")

    print()


# Lower part
for i in range(n - 2, -1, -1):
    print("" * i, end="")

    for j in range(i, 2 * n - 1 - i):
        print(chr(65 + j), end=" ")

    print()
```
