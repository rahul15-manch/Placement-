```python
def reversewordcount(s):
    words=s.split() #["Salesforce","Developer"]
    res=" "
    for word in words:
        word=word[::-1]+str(len(word))
        res+= word+" "
    return res

print(reversewordcount("Salesforce Developer"))
        
```

```python
def armstrong(num):
    temp=num
    ans=0
    for i in range(len(str(temp))):
        d=num%10
        ans=ans+d**(len(str(temp)))
        num//=10
    return temp==ans
print(armstrong(153))

def generatearmstrong(k):

    start= 10**(k-1)
    end=10 **(k)
    for num in range(start,end):
        if armstrong(num)==True:
            return num
        
print(generatearmstrong(5))
```

```python
def reverse(word):
    left=0
    right=len(word)-1
    word=list(word)
    while left<right:
        word[left],word[right]=word[right],word[left]
        left+=1
        right-=1
    return "".join(word)
def transformWords(s):
    words=s.split()
    res=""
    for i in range(len(words)):
        if i%2==0:
            word=reverse(words[i])+str(len(words[i]))
            res+=word+" "
        else:
            word=words[i]+str(len(words[i]))
            res+=word+" "
    return res

print(transformWords("Cloud Analogy Developer"))

```

```python
def check_prime(n):
    prime=True
    if n<2:
        prime=False
    for i in range(2,n//2+1):
        if n%i==0:
            prime=False
            break
    return prime

def give_prime(start,end):
    res=[]
    for i in range(start,end):
        if check_prime(i):
            res.append(i)
    return res[::2]

print(give_prime(1,300))
```

```python
def check_composite(n):
    comp=False
    if n<4:
        comp=False
    for i in range(2,n//2+1):
        if n%i==0:
            comp=True
       

    return comp
def print_comp(start,end):
    res=[]
    for i in range(start,end):
        if check_composite(i):
            res.append(i)
    return res

print(print_comp(1,100))
```
