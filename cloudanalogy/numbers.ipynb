```python
def generatePassword(a, b):

    total = 0

    for num in range(a, b + 1):

        # Check whether num is prime
        if num < 2:
            continue

        prime = True

        i = 2

        while i * i <= num:
            if num % i == 0:
                prime = False
                break
            i += 1

        # If prime, calculate sum of its digits
        if prime:
            temp = num

            while temp > 0:
                digit = temp % 10
                total += digit
                temp //= 10

    return total


print(generatePassword(1, 20))
```

```python
def armstrong(num):
    temp=num
    ans=0
    for i in range(len(str(temp))):
        d=num%10
        ans=ans+d**(len(str(temp)))
        num//=10
    return temp==ans

print(armstrong(153))
```

```python
def armstrong(num):
    temp=num
    ans=0
    for i in range(len(str(temp))):
        d=num%10
        ans=ans+d**(len(str(temp)))
        num//=10
    return temp==ans


def generatearmstrong(k):
    if k<=2:
        return 0
    start=10**(k-1)
    end=10**(k)

    for num in range(start,end):
        if armstrong(num):
            return num 
    return -1
print(generatearmstrong(4))

```
