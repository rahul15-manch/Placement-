```python
arr=[2,5,4,0,8,9]
def second_maxelement(nums):
    first_max=float("-inf")
    second_max=float("-inf")
    for num in nums:
        if num>first_max:
            second_max = first_max
            first_max = num

        elif num>second_max and num!=first_max:
            second_max=num
    return second_max
print(second_maxelement(arr))
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
class LinkedList:
    def __init__(self):
        self.head = None
ll = LinkedList()

ll.head = Node(10)
ll.head.next = Node(20)
ll.head.next.next = Node(30)
```

```python
current = ll.head

while current:
    print(current.data)
    current = current.next
```
